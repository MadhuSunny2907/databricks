```python
from delta.tables import DeltaTable
from pyspark.sql.functions import col
```

```python
#Merge customer data
src_cust_df=spark.table('olist.silver.customers')
cust_target=DeltaTable.forName(spark,'olist.silver.customers_data')
cust_target.alias("t").merge(src_cust_df.alias("s"),"t.customer_id=s.customer_id").whenMatchedUpdate(set={
    "t.customer_unique_id": "s.customer_unique_id",
    "t.customer_zip_code_prefix": "s.customer_zip_code_prefix",
    "t.customer_city": "s.customer_city",
    "t.customer_state": "s.customer_state",
    "t.run_date": "s.run_date",
    "t.created_at": "s.created_at",
    "t.Is_Cust_Repeat": "s.Is_Cust_Repeat",
    "t.updated_tms": "current_timestamp"
}).whenNotMatchedInsert(values={
    "t.customer_id":"s.customer_id",
    "t.customer_unique_id": "s.customer_unique_id", 
    "t.customer_zip_code_prefix": "s.customer_zip_code_prefix",
    "t.customer_city": "s.customer_city",
    "t.customer_state": "s.customer_state",
    "t.run_date": "s.run_date",
    "t.created_at": "s.created_at",
    "t.Is_Cust_Repeat": "s.Is_Cust_Repeat",
    "t.isrt_tms": "current_timestamp"
}).execute()
```

```python
#Merge orders data
src_orders_df=spark.table('olist.silver.orders')
src_cust_data_df=spark.table('olist.silver.customers_data').select('customer_id','cust_key')
src_orders_df=src_orders_df.join(src_cust_data_df,on='customer_id',how='inner')
orders_target_df=DeltaTable.forName(spark,'olist.silver.orders_data')
orders_target_df.alias("t").merge(src_orders_df.alias("s"),"t.order_id=s.order_id").whenMatchedUpdate(set={
    "t.cust_key":"s.cust_key",
    "t.order_status": "s.order_status",
    "t.order_purchase_timestamp": "s.order_purchase_timestamp",
    "t.order_approved_at": "s.order_approved_at",
    "t.order_delivered_carrier_date": "s.order_delivered_carrier_date",
    "t.order_delivered_customer_date": "s.order_delivered_customer_date",
    "t.order_estimated_delivery_date": "s.order_estimated_delivery_date",
    "t.run_date":"s.run_date",
    "t.created_at": "s.created_at",
    "t.late_delivery": "s.late_delivery",
    "t.delivery_days_estimated": "s.delivery_days_estimated",
    "t.delivery_days_actual": "s.delivery_days_actual",
    "t.delay_in_delivery": "s.delay_in_delivery",
    "t.order_year": "s.order_year",
    "t.updated_tms": "current_timestamp"
}).whenNotMatchedInsert(values={
    "t.order_id":"s.order_id",
    "t.cust_key":"s.cust_key",
    "t.order_status": "s.order_status",
    "t.order_purchase_timestamp": "s.order_purchase_timestamp",
    "t.order_approved_at": "s.order_approved_at",
    "t.order_delivered_carrier_date": "s.order_delivered_carrier_date",
    "t.order_delivered_customer_date":"s.order_delivered_customer_date",
    "t.order_estimated_delivery_date": "s.order_estimated_delivery_date",
    "t.run_date":"s.run_date",
    "t.created_at": "s.created_at",
    "t.late_delivery": "s.late_delivery",
    "t.delivery_days_estimated": "s.delivery_days_estimated",
    "t.delivery_days_actual": "s.delivery_days_actual",
    "t.delay_in_delivery": "s.delay_in_delivery",
    "t.order_year": "s.order_year",
    "t.isrt_tms":"current_timestamp"
}).execute()
```

```python
#Merge Seller Data
src_seller_df=spark.table('olist.silver.seller')
tgt_seller_df=DeltaTable.forName(spark,'olist.silver.seller_data')
tgt_seller_df.alias('t').merge(src_seller_df.alias('s'),'t.seller_id=s.seller_id').whenMatchedUpdate(set={
    "t.seller_zip_code_prefix": "s.seller_zip_code_prefix",
    "t.seller_city": "s.seller_city",
    "t.seller_state": "s.seller_state",
    "t.run_date": "s.run_date",
    "t.created_at": "s.created_at",
    "t.updated_tms": "current_timestamp"
}).whenNotMatchedInsert(values={
    "t.seller_id": "s.seller_id",
    "t.seller_zip_code_prefix": "s.seller_zip_code_prefix",
    "t.seller_city": "s.seller_city",
    "t.seller_state": "s.seller_state",
    "t.run_date": "s.run_date",
    "t.created_at": "s.created_at",
    "t.isrt_tms": "current_timestamp"
}).execute()
```

```python
#Merge OrderitemsData
src_orderitems_df=spark.table('olist.silver.orderitems')
src_orders=spark.table('olist.silver.orders_data').select('order_id','order_key')
src_prod_data_df=spark.table('olist.silver.products_data').select('product_id','product_key')
src_seller_data_df=spark.table('olist.silver.seller_data').select('seller_id','seller_key')
src_orderitems_df=src_orderitems_df.join(src_orders,on="order_id",how="inner").join(src_seller_data_df,on="seller_id",how="left").join(src_prod_data_df,on="product_id",how="left")
orderitems_target=DeltaTable.forName(spark,'olist.silver.orderitems_data')
orderitems_target.alias("t").merge(src_orderitems_df.alias("s"), (col("t.order_key") == col("s.order_key")) & (col("t.order_item_id") == col("s.order_item_id"))).whenMatchedUpdate(set={
    "t.product_key": "s.product_key",
    "t.seller_key": "s.seller_key",
    "t.shipping_limit_date":"s.shipping_limit_date",
    "t.price":"s.price",
    "t.freight_value":"s.freight_value",
    "t.run_date":"s.run_date",
    "t.created_at":"s.run_date",
    "t.total_price":"s.total_price",
    "t.freight_pct":"s.freight_pct",
    "t.is_high_freight":"s.is_high_freight",
    "t.updated_tms":"current_timestamp"})\
    .whenNotMatchedInsert(values={
    "t.order_item_id":"s.order_item_id",
    "t.order_key": "s.order_key",
    "t.product_key": "s.product_key",
    "t.seller_key": "s.seller_key",
    "t.shipping_limit_date":"s.shipping_limit_date",
    "t.price":"s.price",
    "t.freight_value":"s.freight_value",
    "t.run_date":"s.run_date",
    "t.created_at":"s.run_date",
    "t.total_price":"s.total_price",
    "t.freight_pct":"s.freight_pct",
    "t.is_high_freight":"s.is_high_freight",
    "t.isrt_tms":"current_timestamp"
        }).execute()
```

```python
#Merge Payments Data
src_payments_df=spark.table('olist.silver.payments')
src_order_data_df=spark.table('olist.silver.orders_data').select('order_id','order_key')
src_payments_df=src_payments_df.join(src_order_data_df,on="order_id",how="inner")
tgt_payments_df=DeltaTable.forName(spark,'olist.silver.payments_data')
tgt_payments_df.alias('t').merge(src_payments_df.alias('s'),'t.order_key=s.order_key').whenMatchedUpdate(set={
    "t.payment_sequential": "s.payment_sequential",
    "t.payment_type": "s.payment_type",
    "t.payment_installments": "s.payment_installments",
    "t.payment_value": "s.payment_value",
    "t.run_date": "s.run_date",
    "t.created_at": "s.created_at",
    "t.updated_tms": "current_timestamp",
    "t.total_payment":"s.total_payment",
    "t.via_installments":"s.via_installments"})\
    .whenNotMatchedInsert(values={
    "t.payment_sequential": "s.payment_sequential",
    "t.payment_type": "s.payment_type",
    "t.payment_installments": "s.payment_installments",
    "t.payment_value": "s.payment_value",
    "t.run_date": "s.run_date",
    "t.created_at": "s.created_at",
    "t.isrt_tms": "current_timestamp",
    "t.total_payment":"s.total_payment",
    "t.via_installments":"s.via_installments"}).execute()
```
