```python
run_date=dbutils.jobs.taskValues.get(taskKey="RUN_DATE", key="run_date", debugValue="20161004")
```

```python
from pyspark.sql.functions import lit,current_timestamp,to_date
#Read and Ingest Customer Data
Cust_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/customer_' + run_date + '.csv')
Cust_df=Cust_df.withColumn('run_date',to_date(lit(run_date), 'yyyyMMdd'))\
    .withColumn('created_at',current_timestamp())
Cust_df.writeTo('olist.bronze.customer').createOrReplace()

#Read and Ingest Orders Data
Orders_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/orders_' + run_date + '.csv')
Orders_df=Orders_df.withColumn('run_date',to_date(lit(run_date), 'yyyyMMdd'))\
    .withColumn('created_at',current_timestamp())
Orders_df.writeTo('olist.bronze.orders').createOrReplace()

#Read and Ingest OrderItems Data
OrderItems_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/orderitems_' + run_date + '.csv')
OrderItems_df=OrderItems_df.withColumn('run_date',to_date(lit(run_date), 'yyyyMMdd'))\
    .withColumn('created_at',current_timestamp())
OrderItems_df.writeTo('olist.bronze.orderItems').createOrReplace()

#Read and Ingest Payments Data
Payments_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/payments_' + run_date + '.csv')
Payments_df=Payments_df.withColumn('run_date',to_date(lit(run_date), 'yyyyMMdd'))\
    .withColumn('created_at',current_timestamp())
Payments_df.writeTo('olist.bronze.payments').createOrReplace()

#Read and Ingest Review Data
Reviews_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/reviews_' + run_date + '.csv')
Reviews_df=Reviews_df.withColumn('run_date',to_date(lit(run_date), 'yyyyMMdd'))\
    .withColumn('created_at',current_timestamp())
Reviews_df.writeTo('olist.bronze.reviews').createOrReplace()

#Read and Ingest Products Data
product_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/products_' + run_date + '.csv')
product_df=product_df.withColumn('run_date',to_date(lit(run_date), 'yyyyMMdd'))\
    .withColumn('created_at',current_timestamp())
product_df.writeTo('olist.bronze.products').createOrReplace()

#Read and Ingest Seller Data
seller_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/sellers_' + run_date + '.csv')
seller_df=seller_df.withColumn('run_date',to_date(lit(run_date), 'yyyyMMdd'))\
    .withColumn('created_at',current_timestamp())
seller_df.writeTo('olist.bronze.seller').createOrReplace()
```
