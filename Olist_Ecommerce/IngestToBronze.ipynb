```python
#Read FlowDate
run_date_df=spark.sql("select to_char(NEXT_PROC_TMS,'yyyyMMdd') as NEXT_PROC_TMS FROM olist.default.flow_status")
run_date=str(run_date_df.collect()[0]['NEXT_PROC_TMS'])
```

```python
#Read and Ingest Customer Data
Cust_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/customer_' + run_date + '.csv')
Cust_df.write.mode('overWrite').saveAsTable('olist.bronze.customer')

#Read and Ingest Orders Data
Orders_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/orders_' + run_date + '.csv')
Orders_df.write.mode('overWrite').saveAsTable('olist.bronze.orders')

#Read and Ingest OrderItems Data
OrderItems_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/orderitems_' + run_date + '.csv')
OrderItems_df.write.mode('overWrite').saveAsTable('olist.bronze.orderItems')

#Read and Ingest Payments Data
Payments_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/payments_' + run_date + '.csv')
Payments_df.write.mode('overWrite').saveAsTable('olist.bronze.payments')

#Read and Ingest Review Data
Reviews_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/reviews_' + run_date + '.csv')
Reviews_df.write.mode('overWrite').saveAsTable('olist.bronze.reviews')

#Read and Ingest Products Data
product_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/products_' + run_date + '.csv')
product_df.write.mode('overWrite').saveAsTable('olist.bronze.products')

#Read and Ingest Seller Data
seller_df=spark.read.option('header','true').option('inferSchema','true').csv('/Volumes/olist/landing/v_olist/sellers_' + run_date + '.csv')
seller_df.write.mode('overWrite').saveAsTable('olist.bronze.seller')
```
