```python
%python
df=spark.table('gizmobox.bronze.refunds')
display(df)
```

```python
from pyspark.sql.functions import split, date_format
df_split=df.select(
    'refund_id',
    'payment_id',
    date_format('refund_timestamp','yyyy-MM-dd').alias('refnud_date'),
    date_format('refund_timestamp','HH:mm:ss').alias('refnud_time'),    
    'refund_amount',
    split('refund_reason',':')[0].alias('payment_reason'),
    split('refund_reason',':')[1].alias('payment_src'))

display(df_split)
df_split.write.mode('overwrite').saveAsTable('gizmobox.silver.py_refunds')
```
